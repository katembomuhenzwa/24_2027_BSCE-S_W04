```python
STUDENT_ID="24/2027/BSCE-S"
COHORT="CML3112"
import csv
import collections
with open("site_inspection_log.csv", newline="") as f:
    rows = list(csv.DictReader(f))
```

## valid_count (5 marks)
Store an integer count of rows whose crack_width_mm field is nonblank. Skip blank readings before numeric conversion.


```python
valid_count = sum(1 for row in rows if row["crack_width_mm"].strip())
valid_count
```

## mean_reading (5 marks)
Store the arithmetic mean of the nonblank crack_width_mm values as a number. Add the valid readings and divide by their count. Keep full precision.


```python
valid_readings = [
    float(row["crack_width_mm"])
    for row in rows
    if row["crack_width_mm"].strip()
]
mean_reading = sum(valid_readings) / valid_count
mean_reading
```

## above_limit (5 marks)
Store an integer count of valid crack_width_mm readings strictly greater than 0.3. A reading equal to the threshold is not counted.


```python
above_limit = sum(1 for value in valid_readings if value > 0.3)
above_limit
```

## group_means (5 marks)
Store a dictionary whose keys are the structure_id labels and whose values are the mean valid crack_width_mm readings within each group. Divide each group total by its own valid count. Keep full precision.


```python
group_totals = collections.defaultdict(float)
group_counts = collections.Counter()
for row in rows:
    reading = row["crack_width_mm"].strip()
    if reading:
        structure_id = row["structure_id"]
        group_totals[structure_id] += float(reading)
        group_counts[structure_id] += 1

group_means = {
    structure_id: group_totals[structure_id] / group_counts[structure_id]
    for structure_id in group_totals
}
group_means
```

### Pattern interpretation (formative practice)
The mean crack width rises consistently from structure B1 to B12, from about 0.111 mm to 0.344 mm. The overall valid-reading mean is about 0.226 mm, and 59 valid readings are strictly above the teaching threshold of 0.3 mm.
